```python
import pandas as pd
import spacy
import re

# Carichiamo un modello NLP avanzato (Transformer-based pipeline)
# Per l'inglese (come nel paper) useremmo 'en_core_web_trf'
# Per l'italiano useremmo 'it_core_news_lg' o simili
try:
    nlp = spacy.load("en_core_web_sm") # Uso sm per leggerezza, in prod usare 'trf'
except:
    print("Modello non trovato. Esegui: python -m spacy download en_core_web_sm")
    # Fallback semplice se spaCy non è installato
    class MockNLP:
        def __call__(self, text):
            return type('Doc', (object,), {'sents': [type('Span', (object,), {'text': s.strip()}) for s in text.split('.') if s]})()
    nlp = MockNLP()

def preprocess_and_split(raw_text):
    """
    Replica la logica descritta nel paper:
    1. Pulizia formattazione (bullet points, virgole) 
    2. Segmentazione basata su modelli linguistici 
    """
    if not isinstance(raw_text, str):
        return []

    # 1. Pre-processing specifico (Cleaning) 
    # Sostituisce i bullet points o le virgole a fine riga di elenco con punti
    # per evitare "excessively long sentences".
    clean_text = raw_text.replace('\r', ' ').replace('\n', ' ')
    # Esempio euristico: se c'è un bullet point (•, -, *) seguito da testo e virgola, normalizziamo
    # Questa regex è un'approssimazione della logica descritta
    clean_text = re.sub(r'\s*[,;]\s*(?=[A-Z•\-])', '. ', clean_text)
    
    # Rimuove spazi multipli
    clean_text = re.sub(r'\s+', ' ', clean_text).strip()

    # 2. Segmentazione Transformer-based 
    doc = nlp(clean_text)
    
    # Estraiamo le frasi (sentences)
    sentences = [sent.text.strip() for sent in doc.sents if len(sent.text.strip()) > 5]
    
    return sentences

# --- ESEMPIO DI UTILIZZO ---
# Simuliamo una riga dal CSV come descritto in Tabella 1 [cite: 155]
sample_syllabus = """
The main security properties of an information system will be introduced and discussed in detail. The approaches for improving security at the various levels of the network protocol stack will then be presented and analyzed. The course adopts an “offensive security” approach. The concepts related to the so-called vulnerability–threat–attack chain will be illustrated. The techniques for preparing a cyber attack, such as footprinting, scanning, and enumeration, will be described. Finally, the final phase of an attack, known as exploitation, will be presented. Topics such as firewalling, intrusion detection, malware analysis, and protection against Distributed Denial of Service (DDoS) attacks will be introduced. Finally, the main techniques of so-called ethical hacking will be presented and analyzed in detail.

Syllabus:

Network security: principles and architecture
- Functional security requirements
- Threats, attacks, countermeasures

Wireless network security

Network-layer security
The IPsec protocol suite

Transport-layer security
- Secure Socket Layer (SSL), Transport Layer Security (TLS), Secure Shell (SSH)

Application-layer security
- Email
- Web
    - HTTPS
    - Security architecture of WebRTC (Web Real-Time Communications)

Cloud computing and security (overview)

Introduction to malware
- Taxonomy
    - Viruses, worms, Trojans, rootkits, etc.
- Advanced Persistent Threats (APTs)
- Countermeasures

Denial of Service (DoS) and Distributed Denial of Service (DDoS) attacks

Intrusion Detection Systems (IDS)
- Host-based, network-based, and hybrid techniques

Firewalls and Intrusion Prevention Systems (IPS)

Hacking in IP networks
- Preliminary phases of an attack
    - Footprinting, scanning, enumeration

- Targeted attack techniques
    - Against end systems and servers
    - Against infrastructure
        - VoIP (Voice over IP) networks
        - Wireless networks
        - Hardware systems
    - Against data and applications
        - Web
        - Mobile devices
        - Databases
"""

sentences = preprocess_and_split(sample_syllabus)

print(f"INPUT:\n{sample_syllabus}\n")
print(f"OUTPUT (Segmented Sentences - {len(sentences)}):")
for i, s in enumerate(sentences):
    print(f"{i+1}: {s}")
```

```python
from sentence_transformers import SentenceTransformer
import numpy as np

def compute_syllabus_embeddings(sentences_list, model_name='all-mpnet-base-v2'):
    """
    Genera gli embedding per una lista di frasi usando un modello Transformer.
    [cite_start]Replica la fase 'Embeddings Preprocessing' del paper[cite: 108].
    
    Args:
        sentences_list (list): Lista di stringhe (frasi del syllabus).
        model_name (str): Nome del modello HuggingFace da utilizzare.
                          'all-mpnet-base-v2' è raccomandato per performance/qualità.
    
    Returns:
        numpy.ndarray: Matrice degli embedding (n_frasi, dimensione_vettore).
    """
    print(f"Loading model '{model_name}'...")
    # Carica il modello (viene scaricato solo la prima volta)
    embedder = SentenceTransformer(model_name)
    
    print(f"Encoding {len(sentences_list)} sentences...")
    # Genera gli embedding. convert_to_numpy=True restituisce array numpy invece di tensor
    embeddings = embedder.encode(sentences_list, convert_to_numpy=True, show_progress_bar=True)
    
    return embeddings

# --- ESEMPIO DI UTILIZZO (Collegato alla cella precedente) ---
# Assicurati che la variabile 'sentences' esista dalla cella precedente
if 'sentences' in locals() and sentences:
    syllabus_embeddings = compute_syllabus_embeddings(sentences)
    
    print(f"\nGenerazione completata.")
    print(f"Shape degli embedding: {syllabus_embeddings.shape}")
    print(f"(Righe: {syllabus_embeddings.shape[0]} frasi, Colonne: {syllabus_embeddings.shape[1]} dimensioni del vettore)")
else:
    print("Errore: La variabile 'sentences' non è definita o è vuota. Esegui prima la cella di splitting.")
```

```python
from sentence_transformers import SentenceTransformer
import numpy as np

def compute_all_ecsf_embeddings(profiles_list, model_name='all-mpnet-base-v2'):
    """
    Calcola gli embedding per TUTTI i ruoli ECSF presenti nella lista.
    
    Args:
        profiles_list (list): Lista di dizionari ECSF (come definita in ECSF_PROFILES).
        model_name (str): Nome del modello SentenceTransformer.
        
    Returns:
        dict: Un dizionario strutturato come:
              {
                  "Nome Ruolo": {
                      "skills_embeddings": array,
                      "knowledge_embeddings": array
                  },
                  ...
              }
    """
    
    # 1. Caricamento Modello (UNA SOLA VOLTA)
    print(f"Loading model '{model_name}'...")
    model = SentenceTransformer(model_name)
    
    all_results = {} # Contenitore per tutti i risultati

    # 2. Iterazione su ogni profilo della lista
    print(f"Processing {len(profiles_list)} ECSF profiles...")
    
    for role_data in profiles_list:
        role_name = role_data['role_name']
        
        # Calcolo Embedding Skills
        skills_text = role_data.get('key_skills', [])
        if skills_text:
            skills_embeddings = model.encode(skills_text, convert_to_numpy=True, show_progress_bar=False)
        else:
            skills_embeddings = np.array([])

        # Calcolo Embedding Knowledge
        knowledge_text = role_data.get('key_knowledge', [])
        if knowledge_text:
            knowledge_embeddings = model.encode(knowledge_text, convert_to_numpy=True, show_progress_bar=False)
        else:
            knowledge_embeddings = np.array([])
            
        # Salviamo nel dizionario globale usando il nome del ruolo come chiave
        all_results[role_name] = {
            "skills_embeddings": skills_embeddings,
            "knowledge_embeddings": knowledge_embeddings
        }
        
    return all_results

# --- ESECUZIONE ---
# Passiamo la lista completa ECSF_PROFILES creata nello step precedente
ecsf_embeddings_dict = compute_all_ecsf_embeddings(ECSF_PROFILES, model_name='all-mpnet-base-v2')

# --- VERIFICA ---
# Stampiamo un esempio per vedere se ha funzionato
test_role = "Chief Information Security Officer (CISO)"

if test_role in ecsf_embeddings_dict:
    print(f"\n--- Risultato Calcolo per: {test_role} ---")
    s_shape = ecsf_embeddings_dict[test_role]['skills_embeddings'].shape
    k_shape = ecsf_embeddings_dict[test_role]['knowledge_embeddings'].shape
    
    print(f"Skills Matrix Shape: {s_shape}")       # Es: (16, 768)
    print(f"Knowledge Matrix Shape: {k_shape}")    # Es: (11, 768)
else:
    print("Errore: Ruolo non trovato.")


```

```python
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

def calculate_ecsf_coverage(syllabus_embeddings, ecsf_dict, threshold=0.51):
    """
    Calcola il punteggio di copertura per ogni ruolo ECSF confrontando gli embedding.
    Replica la logica CSCAM descritta nel paper.
    
    Args:
        syllabus_embeddings (np.array): Matrice embedding del syllabus (N_frasi, 768).
        ecsf_dict (dict): Dizionario output di compute_all_ecsf_embeddings.
        threshold (float): Soglia di similarità minima (0.51 come da paper).
        
    Returns:
        dict: Dizionario { "Nome Ruolo": score_percentuale }
        dict: Dizionario dettagliato con i match specifici (opzionale per debug)
    """
    
    role_scores = {}
    detailed_matches = {} # Per vedere quali frasi hanno attivato la copertura
    
    print(f"--- Starting Coverage Analysis (Threshold: {threshold}) ---")
    
    for role_name, data in ecsf_dict.items():
        skills = data['skills_embeddings']
        knowledge = data['knowledge_embeddings']
        
        # 1. Uniamo Skills e Knowledge in un unico set di "Requisiti del Ruolo"
        # Gestiamo il caso in cui uno dei due sia vuoto
        if skills.size > 0 and knowledge.size > 0:
            role_requirements = np.vstack([skills, knowledge])
        elif skills.size > 0:
            role_requirements = skills
        elif knowledge.size > 0:
            role_requirements = knowledge
        else:
            print(f"Skipping {role_name}: No embeddings found.")
            role_scores[role_name] = 0.0
            continue
            
        # 2. Calcolo Matrice di Similarità (Requisiti vs Syllabus)
        # Shape risultante: (N_Requisiti, N_Frasi_Syllabus)
        # Ogni riga rappresenta un requisito ECSF, ogni colonna una frase del syllabus
        similarity_matrix = cosine_similarity(role_requirements, syllabus_embeddings)
        
        # 3. Trova il "Best Match" per ogni requisito
        # Per ogni riga (requisito), prendiamo il valore massimo (la frase del syllabus più simile)
        max_similarities = similarity_matrix.max(axis=1)
        
        # 4. Applica la Soglia (Threshold Check)
        # Contiamo quanti requisiti hanno trovato almeno una frase con similarità > 0.51
        covered_count = np.sum(max_similarities >= threshold)
        total_requirements = role_requirements.shape[0]
        
        # 5. Calcolo Percentuale
        coverage_percentage = (covered_count / total_requirements) * 100
        role_scores[role_name] = round(coverage_percentage, 4)
        
        # (Opzionale) Debug info
        detailed_matches[role_name] = {
            "total_reqs": total_requirements,
            "covered": covered_count,
            "avg_max_sim": round(np.mean(max_similarities), 4)
        }

    return role_scores, detailed_matches

# --- ESECUZIONE ---

# Calcoliamo i punteggi
scores, details = calculate_ecsf_coverage(syllabus_embeddings, ecsf_embeddings_dict, threshold=0.51)

# Stampiamo i risultati in formato tabellare
print("\n" + "="*60)
print(f"{'ECSF ROLE':<50} | {'SCORE (%)':>9}")
print("="*60)

# Ordiniamo per score decrescente per leggibilità
for role, score in sorted(scores.items(), key=lambda x: x[1], reverse=True):
    print(f"{role:<50} | {score:>9.2f}")
print("="*60)
```

```python
import json
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

# --- 1. FUNZIONE DI CALCOLO COPERTURA (Invariata) ---
def calculate_ecsf_coverage(syllabus_embeddings, ecsf_dict, threshold=0.51):
    """
    Calcola il punteggio di copertura per ogni ruolo ECSF confrontando gli embedding.
    """
    role_scores = {}
    
    print(f"--- Starting Coverage Analysis (Threshold: {threshold}) ---")
    
    for role_name, data in ecsf_dict.items():
        skills = data['skills_embeddings']
        knowledge = data['knowledge_embeddings']
        
        # Unione Skills e Knowledge
        if skills.size > 0 and knowledge.size > 0:
            role_requirements = np.vstack([skills, knowledge])
        elif skills.size > 0:
            role_requirements = skills
        elif knowledge.size > 0:
            role_requirements = knowledge
        else:
            role_scores[role_name] = 0.0
            continue
            
        # Calcolo Similarità Coseno (Requisiti vs Syllabus)
        similarity_matrix = cosine_similarity(role_requirements, syllabus_embeddings)
        
        # Best Match per ogni requisito
        max_similarities = similarity_matrix.max(axis=1)
        
        # Soglia e Percentuale
        covered_count = np.sum(max_similarities >= threshold)
        total_requirements = role_requirements.shape[0]
        
        coverage_percentage = (covered_count / total_requirements) * 100
        role_scores[role_name] = round(coverage_percentage, 4)

    return role_scores

# --- 2. ESECUZIONE DEL CALCOLO ---
# Assumiamo che 'syllabus_embeddings' e 'ecsf_embeddings_dict' siano già stati calcolati
scores = calculate_ecsf_coverage(syllabus_embeddings, ecsf_embeddings_dict, threshold=0.51)

# --- 3. FORMATTAZIONE E SALVATAGGIO JSON ---
# Definiamo le chiavi esatte richieste per garantire l'ordine e la struttura
target_keys = [
    "Chief Information Security Officer (CISO)",
    "Cyber Incident Responder",
    "Cyber Legal, Policy & Compliance Officer",
    "Cyber Threat Intelligence Specialist",
    "Cybersecurity Architect",
    "Cybersecurity Auditor",
    "Cybersecurity Educator",
    "Cybersecurity Implementer",
    "Cybersecurity Researcher",
    "Cybersecurity Risk Manager",
    "Digital Forensics Investigator",
    "Penetration Tester"
]

# Creiamo il dizionario ordinato mappando i risultati calcolati
json_output_data = {}

print("\n--- Generating JSON Output ---")
for key in target_keys:
    # Recuperiamo lo score calcolato; se per qualche motivo manca, mettiamo 0.0
    valore = scores.get(key, 0.0)
    json_output_data[key] = valore
    print(f"Mapped {key[:30]}... -> {valore}")

# Salvataggio su file
output_filename = "computed_coverage.json"
with open(output_filename, "w") as f:
    json.dump(json_output_data, f, indent=4)

print(f"\n[SUCCESS] File JSON salvato come: {output_filename}")

# Stampa di verifica del contenuto JSON
print("\nContenuto JSON generato:")
print(json.dumps(json_output_data, indent=4))
```
