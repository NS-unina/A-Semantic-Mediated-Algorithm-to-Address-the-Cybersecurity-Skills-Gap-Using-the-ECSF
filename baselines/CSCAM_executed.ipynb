## Reproducibility note

The historical CSCAM reimplementation used spaCy sentence segmentation.
The saved notebook output that produced the manuscript results contained
**13 syllabus segments**. Because spaCy model availability/version can alter
segmentation, this reproducibility version freezes those exact 13 segments
below.

No CSCAM scoring rule is changed: the model remains
`all-mpnet-base-v2`, the similarity measure remains cosine similarity, and the
coverage threshold remains `0.51`.


```python
sentences = [
    "The main security properties of an information system will be introduced and discussed in detail.",
    "The approaches for improving security at the various levels of the network protocol stack will then be presented and analyzed.",
    "The course adopts an “offensive security” approach.",
    "The concepts related to the so-called vulnerability–threat–attack chain will be illustrated.",
    "The techniques for preparing a cyber attack, such as footprinting, scanning, and enumeration, will be described.",
    "Finally, the final phase of an attack, known as exploitation, will be presented.",
    "Topics such as firewalling, intrusion detection, malware analysis, and protection against Distributed Denial of Service (DDoS) attacks will be introduced.",
    "Finally, the main techniques of so-called ethical hacking will be presented and analyzed in detail.",
    "Syllabus: Network security: principles and architecture - Functional security requirements - Threats, attacks, countermeasures Wireless network security Network-layer security The IPsec protocol suite Transport-layer security - Secure Socket Layer (SSL).",
    "Transport Layer Security (TLS).",
    "Secure Shell (SSH) Application-layer security - Email - Web - HTTPS - Security architecture of WebRTC (Web Real-Time Communications)",
    "Cloud computing and security (overview) Introduction to malware - Taxonomy - Viruses, worms.",
    "Trojans, rootkits, etc. - Advanced Persistent Threats (APTs) - Countermeasures Denial of Service (DoS) and Distributed Denial of Service (DDoS) attacks Intrusion Detection Systems (IDS) - Host-based, network-based, and hybrid techniques Firewalls and Intrusion Prevention Systems (IPS) Hacking in IP networks - Preliminary phases of an attack - Footprinting, scanning, enumeration - Targeted attack techniques - Against end systems and servers - Against infrastructure - VoIP (Voice over IP) networks - Wireless networks - Hardware systems - Against data and applications - Web - Mobile devices - Databases"
]

assert len(sentences) == 13
print(f"Using {len(sentences)} frozen syllabus segments.")
for i, sentence in enumerate(sentences, start=1):
    print(f"{i}: {sentence}")
```

```python

from sentence_transformers import SentenceTransformer
import numpy as np

MODEL_NAME = "all-mpnet-base-v2"

print(f"Loading model '{MODEL_NAME}'...")
model = SentenceTransformer(MODEL_NAME)

syllabus_embeddings = model.encode(
    sentences,
    convert_to_numpy=True,
    show_progress_bar=True,
)

print("Syllabus embedding shape:", syllabus_embeddings.shape)
assert syllabus_embeddings.shape[0] == 13
```

```python

def compute_all_ecsf_embeddings(profiles_list):
    all_results = {}

    for role_data in profiles_list:
        role_name = role_data["role_name"]

        skills_text = role_data.get("key_skills", [])
        knowledge_text = role_data.get("key_knowledge", [])

        skills_embeddings = (
            model.encode(
                skills_text,
                convert_to_numpy=True,
                show_progress_bar=False,
            )
            if skills_text
            else np.array([])
        )

        knowledge_embeddings = (
            model.encode(
                knowledge_text,
                convert_to_numpy=True,
                show_progress_bar=False,
            )
            if knowledge_text
            else np.array([])
        )

        all_results[role_name] = {
            "skills_embeddings": skills_embeddings,
            "knowledge_embeddings": knowledge_embeddings,
        }

    return all_results


ecsf_embeddings_dict = compute_all_ecsf_embeddings(ECSF_PROFILES)
print(f"Processed {len(ecsf_embeddings_dict)} ECSF profiles.")
```

```python

from sklearn.metrics.pairwise import cosine_similarity

THRESHOLD = 0.51

def calculate_ecsf_coverage(
    syllabus_embeddings,
    ecsf_dict,
    threshold=THRESHOLD,
):
    role_scores = {}
    detailed_matches = {}

    for role_name, data in ecsf_dict.items():
        skills = data["skills_embeddings"]
        knowledge = data["knowledge_embeddings"]

        if skills.size > 0 and knowledge.size > 0:
            role_requirements = np.vstack([skills, knowledge])
        elif skills.size > 0:
            role_requirements = skills
        elif knowledge.size > 0:
            role_requirements = knowledge
        else:
            role_scores[role_name] = 0.0
            continue

        similarity_matrix = cosine_similarity(
            role_requirements,
            syllabus_embeddings,
        )

        max_similarities = similarity_matrix.max(axis=1)
        covered_count = int(
            np.sum(max_similarities >= threshold)
        )
        total_requirements = int(
            role_requirements.shape[0]
        )

        coverage_percentage = (
            covered_count / total_requirements
        ) * 100.0

        role_scores[role_name] = round(
            float(coverage_percentage),
            4,
        )

        detailed_matches[role_name] = {
            "total_requirements": total_requirements,
            "covered_requirements": covered_count,
            "average_max_similarity": round(
                float(np.mean(max_similarities)),
                4,
            ),
        }

    return role_scores, detailed_matches


scores, details = calculate_ecsf_coverage(
    syllabus_embeddings,
    ecsf_embeddings_dict,
    threshold=THRESHOLD,
)

print(f"CSCAM coverage scores (threshold={THRESHOLD}):")
for role, score in sorted(
    scores.items(),
    key=lambda x: x[1],
    reverse=True,
):
    print(f"{role:<50} | {score:9.4f}")
```

```python

from pathlib import Path
import json

ROLE_ORDER = [
    "Chief Information Security Officer (CISO)",
    "Cyber Incident Responder",
    "Cyber Legal, Policy & Compliance Officer",
    "Cyber Threat Intelligence Specialist",
    "Cybersecurity Architect",
    "Cybersecurity Auditor",
    "Cybersecurity Educator",
    "Cybersecurity Implementer",
    "Cybersecurity Researcher",
    "Cybersecurity Risk Manager",
    "Digital Forensics Investigator",
    "Penetration Tester",
]

missing = [role for role in ROLE_ORDER if role not in scores]
if missing:
    raise ValueError(f"Missing CSCAM roles: {missing}")

json_output = {
    role: float(scores[role])
    for role in ROLE_ORDER
}

# Local baseline output.
Path("CSCAM.json").write_text(
    json.dumps(json_output, indent=4),
    encoding="utf-8",
)

# Also place a generated copy in the repository output tree when the notebook
# is executed from baselines/.
generated_dir = Path("../results/generated/baselines")
generated_dir.mkdir(parents=True, exist_ok=True)
(generated_dir / "CSCAM.json").write_text(
    json.dumps(json_output, indent=4),
    encoding="utf-8",
)

print(json.dumps(json_output, indent=4))
print("\n[SUCCESS] CSCAM.json generated.")
```
